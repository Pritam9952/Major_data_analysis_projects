# 📊 Netflix Customer Churn Analysis - EDA

This notebook explores customer behavior to identify churn patterns and generate actionable business insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
sns.set_palette("Set2")
plt.rcParams['figure.figsize'] = (6,4)
```

```python
df = pd.read_csv("../data/processed/cleaned_data.csv")
df.columns = df.columns.str.strip().str.lower()
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
sns.countplot(data=df, x='churn', palette='coolwarm')
plt.title("Customer Churn Distribution", fontweight='bold')
plt.show()
```

### 📊 Insight
- Around 26% customers have churned

### 🔍 Reason
- Moderate churn indicates retention issues

### 💼 Business Meaning
- Focus on retention strategies

```python
sns.countplot(data=df, x='gender', hue='churn')
plt.title("Churn Distribution by Gender", fontweight='bold')
plt.show()
```

### 📊 Insight
- Churn similar across genders

### 🔍 Reason
- Gender not a strong factor

### 💼 Business Meaning
- No gender-based targeting needed

```python
sns.countplot(data=df, x='seniorcitizen', hue='churn')
plt.title("Churn by Senior Citizen Status", fontweight='bold')
plt.show()
```

### 📊 Insight
- Seniors churn more

### 🔍 Reason
- Usability or pricing issues

### 💼 Business Meaning
- Improve senior support

```python
sns.countplot(data=df, x='partner', hue='churn')
plt.title("Churn by Partner Status", fontweight='bold')
plt.show()
```

### 📊 Insight
- Non-partner users churn more

### 🔍 Reason
- Less commitment

### 💼 Business Meaning
- Offer bundled plans

```python
sns.countplot(data=df, x='dependents', hue='churn')
plt.title("Churn by Dependents", fontweight='bold')
plt.show()
```

### 📊 Insight
- No dependents → more churn

### 🔍 Reason
- Less dependency

### 💼 Business Meaning
- Promote family plans

```python
sns.countplot(data=df, x='contract', hue='churn')
plt.title("Churn by Contract Type", fontweight='bold')
plt.show()
```

### 📊 Insight
- Month-to-month highest churn

### 🔍 Reason
- Flexibility to leave

### 💼 Business Meaning
- Push long-term plans

```python
sns.countplot(data=df, x='paymentmethod', hue='churn')
plt.xticks(rotation=30)
plt.title("Churn by Payment Method", fontweight='bold')
plt.show()
```

### 📊 Insight
- Electronic check churn high

### 🔍 Reason
- Less stable payments

### 💼 Business Meaning
- Promote auto-pay

```python
sns.countplot(data=df, x='internetservice', hue='churn')
plt.title("Churn by Internet Service", fontweight='bold')
plt.show()
```

### 📊 Insight
- Fiber users churn more

### 🔍 Reason
- Cost/service issues

### 💼 Business Meaning
- Improve service quality

```python
sns.boxplot(data=df, x='churn', y='monthlycharges')
plt.title("Monthly Charges vs Churn", fontweight='bold')
plt.show()
```

### 📊 Insight
- High charges → high churn

### 🔍 Reason
- Price sensitivity

### 💼 Business Meaning
- Optimize pricing

```python
sns.histplot(data=df, x='tenure', kde=True)
plt.title("Customer Tenure Distribution", fontweight='bold')
plt.show()
```

### 📊 Insight
- Early churn high

### 🔍 Reason
- Poor onboarding

### 💼 Business Meaning
- Improve early experience

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap", fontweight='bold')
plt.show()
```

### 📊 Insight
- MonthlyCharges and tenure show correlation with churn  

### 🔍 Reason
- Pricing and customer duration impact retention  

### 💼 Business Meaning
- Optimize pricing and improve long-term engagement strategies

```python

df['tenure_group'] = pd.cut(
    df['tenure'],
    bins=[0,12,24,48,72],
    labels=['0-1 Year','1-2 Years','2-4 Years','4+ Years']
)

# Plot
sns.countplot(data=df, x='tenure_group', hue='churn')
plt.title("Churn by Tenure Group", fontweight='bold')
plt.show()
```

### 📊 Insight
- Customers in the first year show the highest churn  

### 🔍 Reason
- Early dissatisfaction or poor onboarding experience  

### 💼 Business Meaning
- Improve onboarding and first-year engagement strategies  

### ⚠️ Actionable Step
- Provide welcome offers, support, and engagement programs for new users

## 📌 Key Insights

- High churn in month-to-month, high charges, seniors  
- Low churn in long-term contracts  

## 🚀 Recommendations

- Promote long-term plans  
- Improve onboarding  
- Offer retention discounts
