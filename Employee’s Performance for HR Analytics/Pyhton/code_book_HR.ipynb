## Age Distribution

```python
sns.set_theme(style="white", context="paper", palette="husl")

plt.figure(figsize = (8,5))
df['age'].hist(
    bins = 20,
    edgecolor = 'black',
)
plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
plt.show()
```

## Average Training Score Distribution

```python
plt.figure(figsize=(8,5))
df['avg_training_score'].hist(
    bins = 15,
    color = 'skyblue',
    edgecolor = 'black',
    rwidth = 0.8
)
plt.title('Distribution of Tranning Scores')
plt.xlabel('Average Traning Score')
plt.ylabel('Number of Employees')
plt.show()
```

## Length of Service

```python
plt.figure(figsize=(8,5))
df['length_of_service'].hist(
    bins = 20,
    color = 'orange',
    edgecolor = 'black'
)
plt.title("Length of Service Distribution")
plt.xlabel("Years of Service")
plt.ylabel("Number of Employees")
plt.show()
```

## Department-Wise Employee Count

```python
dept_counts = df['department'].value_counts()

plt.figure(figsize=(10,5))
dept_counts.plot(kind='bar', color="teal", edgecolor="black")
plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=45)
plt.show()
```

## Gender Ratio

```python
gender_counts = df['gender'].value_counts()

plt.figure(figsize=(5,5))
plt.pie(
    gender_counts,
    labels= gender_counts.index,
    autopct='%1.1f%%',
    colors=["#66b3ff","#ff9999"]
)
plt.title("Gender Distribution")
plt.show()
```

## Correlation Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Numeric columns correlation matrix
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Employee Features")
plt.show()
```

## Training Score vs KPI Met

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="KPIs_met_more_than_80",
            y="avg_training_score",
            data = df, palette="Set2",
           hue = 'KPIs_met_more_than_80' )
plt.title("Training Score vs KPI Met (1 = Met KPI)")
plt.xlabel("KPI Met (0 = No, 1 = Yes)")
plt.ylabel("Average Training Score")
plt.show()
```

## Service Length vs KPI Met

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    x="KPIs_met_more_than_80",
    y="length_of_service",
    data = df,
    palette="Set3",
    hue = 'KPIs_met_more_than_80'
)
plt.title("Service Length vs KPI Met (1 = Met KPI)")
plt.xlabel("KPI Met (0 = No, 1 = Yes)")
plt.ylabel("Length of Service (Years)")
plt.show()
```

## Department-wise Performance

```python
import matplotlib.ticker as mtick


dept_perf = df.groupby('department')['KPIs_met_more_than_80'].mean().sort_values()

plt.figure(figsize=(10,5))
dept_perf.plot(kind="barh", color="skyblue", edgecolor="black")
plt.title("Department-wise KPI Met %")
plt.xlabel("KPI Met %")
plt.ylabel("Department")
plt.gca().xaxis.set_major_formatter(mtick.PercentFormatter(1.0))
plt.show()
```

## Education-wise KPI Met %

```python
edu_perf = df.groupby('education')['KPIs_met_more_than_80'].mean().sort_values()

plt.figure(figsize = (8,5))
edu_perf.plot(
    kind = 'bar',
    color = 'skyblue',
    edgecolor = 'black'
)
plt.title('Education-wise KPI Met %')
plt.ylabel('KPI MET (%)')
plt.xlabel('Education Level')
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
plt.show()
```

## Recruitment Channel-wise KPI Met %

```python
rec_perf = df.groupby("recruitment_channel")["KPIs_met_more_than_80"].mean().sort_values() * 100

plt.figure(figsize=(8,5))
rec_perf.plot(kind="bar", color="green", edgecolor="black")
plt.title("Recruitment Channel-wise KPI Met %")
plt.ylabel("KPI Met (%)")
plt.xlabel("Recruitment Channel")
plt.show()
```

## Region-wise KPI Met %

```python
region_perf = df.groupby("region")["KPIs_met_more_than_80"].mean().sort_values() * 100
plt.figure(figsize=(10,6))
region_perf.plot(kind="bar", color="steelblue", edgecolor="black")
plt.title("Region-wise KPI Met %")
plt.ylabel("KPI Met (%)")
plt.xlabel("Region")
plt.xticks(rotation=90)
plt.show()
```

## High Performer vs Low Performer Profile

```python
perf_profile = df.groupby("KPIs_met_more_than_80")[["age", "length_of_service", "avg_training_score"]].mean()

perf_profile.plot(kind="bar", figsize=(8,5), edgecolor="black")
plt.title("High Performer vs Low Performer Profile")
plt.ylabel("Average Values")
plt.xticks([0,1], ["Low Performer", "High Performer"], rotation=0)
plt.show()
```

# 📊 HR Analytics – Python EDA

---

## 🔹 Project Overview
This notebook analyzes employee performance data using Python.  
We explore workforce demographics, training effectiveness, and KPI achievements to derive HR insights.

---

## 🔹 Data Cleaning Steps
- Removed **duplicate records** (2 rows).
- Handled missing values:  
  - `education` column → filled blanks with `"Unknown"`.
- Standardized categorical variables:  
  - `m → Male`, `f → Female`.
- Final dataset shape: **17,415 rows × 13 columns**.

---

## 🔹 Exploratory Data Analysis (EDA)

### 1. Workforce Demographics
- **Age Distribution** → Most employees are between **25–35 years old**.  
- **Length of Service** → Majority have **< 10 years of service**, very few >20 years.  
- **Gender Ratio** → Workforce is **Male-dominated**.  
- **Education Distribution** → Mostly **Bachelors**, followed by **Masters & above**.

---

### 2. Training & Performance
- **Training Score Distribution** → Most employees scored between **50–75**.  
- **Training Score vs KPI Met** → High performers have **consistently higher training scores**.  
- **Service Length vs KPI Met** → Longer service does not guarantee KPI success.  

---

### 3. Departmental Insights
- **Department Size** → Largest workforce in **Operations & Sales**, smallest in **Legal**.  
- **Department-wise KPI %** →  
  - **R&D** shows the **highest KPI met % (~45%)**.  
  - **Sales & Marketing** has the **lowest KPI met % (~28%)**.

---

### 4. Education & Recruitment
- **Education-wise KPI %** → Employees with **Masters** perform slightly better than Bachelors.  
- **Recruitment Channel KPI %** → Referral hires perform **better on average** compared to other channels.

---

### 5. Regional Analysis
- **Region-wise KPI %** → Performance varies significantly by region.  
  Some regions consistently show **lower KPI achievement**, signaling need for targeted HR interventions.

---

### 6. High vs Low Performer Profile
- **High performers** (KPI met >80%):  
  - Higher average **training scores**.  
  - Slightly lower average **service length** compared to low performers.  
- **Low performers**:  
  - Lower training scores.  
  - Longer tenure does not guarantee better KPI outcomes.

---

## 🔹 Key Insights
- Workforce is young (25–35 yrs) and mostly Bachelors-educated.  
- Higher training scores strongly correlate with KPI achievement.  
- Departmental performance varies: R&D leads, Sales & Marketing lags.  
- Recruitment via referrals shows higher-performing employees.  
- Some regions need focused performance improvement programs.
